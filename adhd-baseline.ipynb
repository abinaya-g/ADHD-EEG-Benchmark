```python
import glob
import numpy as np
from scipy.io import loadmat

fs_assumed = 128
epoch_seconds = 30
epoch_len = fs_assumed * epoch_seconds  # 3840 samples

base = "/kaggle/input/datasets/abinayajone/adhd-eeg-dataset"
folders = {
    "ADHD_part1": 1, "ADHD_part2": 1,
    "Control_part1": 0, "Control_part2": 0,
}

all_epochs = []
all_labels = []
all_subject_ids = []

for folder_name, label in folders.items():
    folder_path = f"{base}/{folder_name}/{folder_name}"
    files = sorted(glob.glob(folder_path + "/*.mat"))
    print(f"{folder_name}: {len(files)} files")

    for f in files:
        data = loadmat(f)
        key = [k for k in data.keys() if not k.startswith("__")][0]
        arr = data[key]  # (n_samples, 19)

        n_epochs = arr.shape[0] // epoch_len
        subject_id = f.split("/")[-1].replace(".mat", "")

        for i in range(n_epochs):
            segment = arr[i*epoch_len : (i+1)*epoch_len, :]  # (3840, 19)
            all_epochs.append(segment.T)  # transpose -> (19, 3840)
            all_labels.append(label)
            all_subject_ids.append(subject_id)

X = np.array(all_epochs)[..., np.newaxis]  # (n_epochs, 19, 3840, 1)
y = np.array(all_labels)
groups = np.array(all_subject_ids)

print("\nFinal shapes:")
print("X:", X.shape)
print("y:", y.shape, " -> ADHD epochs:", (y==1).sum(), " Control epochs:", (y==0).sum())
print("groups:", groups.shape, " -> unique subjects:", len(set(groups)))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_cnn(n_channels=19, n_timesamples=3840, fs=128):
    temporal_kernel_1 = int(fs / 4)
    temporal_kernel_2 = int(fs / 8)

    inp = layers.Input(shape=(n_channels, n_timesamples, 1), name="input_eeg")

    x = layers.Conv2D(filters=16, kernel_size=(10, 1), padding="valid",
                       activation="relu", name="spatial_conv1")(inp)
    x = layers.BatchNormalization(name="bn_spatial1")(x)
    x = layers.AveragePooling2D(pool_size=(2, 1), name="pool_spatial1")(x)

    x = layers.Conv2D(filters=16, kernel_size=(4, 1), padding="valid",
                       activation="relu", name="spatial_conv2")(x)
    x = layers.BatchNormalization(name="bn_spatial2")(x)
    x = layers.AveragePooling2D(pool_size=(2, 1), name="pool_spatial2")(x)

    x = layers.Reshape((x.shape[2], x.shape[3]), name="reshape_to_temporal")(x)

    x = layers.Conv1D(filters=32, kernel_size=temporal_kernel_1,
                       padding="valid", activation="relu",
                       name="temporal_conv1")(x)
    x = layers.BatchNormalization(name="bn_temporal1")(x)
    x = layers.AveragePooling1D(pool_size=temporal_kernel_1 // 2,
                                 name="pool_temporal1")(x)

    x = layers.Conv1D(filters=32, kernel_size=temporal_kernel_2,
                       padding="valid", activation="relu",
                       name="temporal_conv2")(x)
    x = layers.BatchNormalization(name="bn_temporal2")(x)
    x = layers.AveragePooling1D(pool_size=temporal_kernel_2 // 2,
                                 name="pool_temporal2")(x)

    flat = layers.Flatten(name="flatten")(x)
    dense1 = layers.Dense(64, activation="relu", name="dense1")(flat)
    dense2 = layers.Dense(32, activation="relu", name="dense2")(dense1)

    out = layers.Dense(1, activation="sigmoid", name="classification")(dense2)

    model = models.Model(inputs=inp, outputs=out, name="adhd_cnn")
    feature_model = models.Model(inputs=inp, outputs=[flat, dense1, dense2],
                                  name="adhd_cnn_features")

    return model, feature_model
```

```python
from sklearn.model_selection import GroupShuffleSplit
import numpy as np

# Quick single split, just to test the pipeline runs (NOT a real evaluation)
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(X, y, groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
print("Train class balance:", np.bincount(y_train))
print("Test class balance:", np.bincount(y_test))

import tensorflow as tf

model, feature_model = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
              loss="binary_crossentropy", metrics=["accuracy"])

history = model.fit(X_train, y_train, epochs=5, batch_size=16, verbose=1,
                     validation_data=(X_test, y_test))
```

```python
import numpy as np

# Normalize each epoch, per channel, using z-score
# (mean/std computed per-channel, within each epoch)
def normalize_epoch(epoch):
    # epoch shape: (19, 3840, 1)
    mean = epoch.mean(axis=1, keepdims=True)
    std = epoch.std(axis=1, keepdims=True)
    std[std == 0] = 1  # avoid divide-by-zero on flat channels
    return (epoch - mean) / std

X_norm = np.array([normalize_epoch(e) for e in X])

print("Before normalization - mean/std of first epoch, channel 0:",
      X[0, 0, :, 0].mean(), X[0, 0, :, 0].std())
print("After normalization  - mean/std of first epoch, channel 0:",
      X_norm[0, 0, :, 0].mean(), X_norm[0, 0, :, 0].std())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

X_train_norm, X_test_norm = X_norm[train_idx], X_norm[test_idx]

model2, feature_model2 = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
model2.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
               loss="binary_crossentropy", metrics=["accuracy"])

history2 = model2.fit(X_train_norm, y_train, epochs=30, batch_size=16, verbose=1,
                       validation_data=(X_test_norm, y_test))

train_feats_norm = feature_model2.predict(X_train_norm, verbose=0)[1]
test_feats_norm = feature_model2.predict(X_test_norm, verbose=0)[1]

lr2 = LogisticRegression(penalty="l2", max_iter=2000)
lr2.fit(train_feats_norm, y_train)

y_pred_norm = lr2.predict(test_feats_norm)
y_proba_norm = lr2.predict_proba(test_feats_norm)[:, 1]

cnn_pred_norm = (model2.predict(X_test_norm, verbose=0).ravel() >= 0.5).astype(int)

print("CNN-only test accuracy (normalized):", accuracy_score(y_test, cnn_pred_norm))
print("LR-hybrid test accuracy (normalized):", accuracy_score(y_test, y_pred_norm))
print("LR-hybrid test AUC (normalized):", roc_auc_score(y_test, y_proba_norm))
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,          # base paper mentions 30-epoch patience; I'm using 10
                            # here since our dataset/epochs are much smaller than
                            # theirs — you may want to tune this
    restore_best_weights=True
)

model3, feature_model3 = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
model3.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
               loss="binary_crossentropy", metrics=["accuracy"])

history3 = model3.fit(X_train_norm, y_train, epochs=100, batch_size=16, verbose=1,
                       validation_data=(X_test_norm, y_test),
                       callbacks=[early_stop])

train_feats3 = feature_model3.predict(X_train_norm, verbose=0)[1]
test_feats3 = feature_model3.predict(X_test_norm, verbose=0)[1]

lr3 = LogisticRegression(penalty="l2", max_iter=2000)
lr3.fit(train_feats3, y_train)

y_pred3 = lr3.predict(test_feats3)
y_proba3 = lr3.predict_proba(test_feats3)[:, 1]
cnn_pred3 = (model3.predict(X_test_norm, verbose=0).ravel() >= 0.5).astype(int)

print("Stopped at epoch:", len(history3.history["loss"]))
print("CNN-only test accuracy:", accuracy_score(y_test, cnn_pred3))
print("LR-hybrid test accuracy:", accuracy_score(y_test, y_pred3))
print("LR-hybrid test AUC:", roc_auc_score(y_test, y_proba3))
```

```python
import itertools

param_grid = {
    "patience": [5, 10],
    "learning_rate": [1e-4, 5e-4],
}

results_grid = []

for patience, lr_val in itertools.product(param_grid["patience"], param_grid["learning_rate"]):
    print(f"\n--- Trying patience={patience}, lr={lr_val} ---")

    early_stop_tune = EarlyStopping(
        monitor="val_loss", patience=patience, restore_best_weights=True
    )

    model_tune, _ = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
    model_tune.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr_val),
        loss="binary_crossentropy", metrics=["accuracy"]
    )

    hist_tune = model_tune.fit(
        X_inner_train, y_inner_train,
        epochs=100, batch_size=16, verbose=0,
        validation_data=(X_inner_val, y_inner_val),
        callbacks=[early_stop_tune]
    )

    best_val_loss = min(hist_tune.history["val_loss"])
    best_val_acc = hist_tune.history["val_accuracy"][
        hist_tune.history["val_loss"].index(best_val_loss)
    ]
    n_epochs_run = len(hist_tune.history["loss"])

    results_grid.append({
        "patience": patience, "lr": lr_val,
        "best_val_loss": best_val_loss, "best_val_acc": best_val_acc,
        "epochs_run": n_epochs_run
    })
    print(f"  best_val_loss={best_val_loss:.4f}  best_val_acc={best_val_acc:.4f}  epochs_run={n_epochs_run}")

print("\n=== Grid search summary ===")
for r in results_grid:
    print(r)
```

```python
from sklearn.model_selection import GroupKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from tensorflow.keras.callbacks import EarlyStopping
import numpy as np

def compute_metrics(y_true, y_pred, y_proba):
    return {
        "accuracy": accuracy_score(y_true, y_pred) * 100,
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "auc": roc_auc_score(y_true, y_proba) if len(set(y_true)) > 1 else np.nan,
    }

n_folds = 10
gkf = GroupKFold(n_splits=n_folds)

fold_results = {"CNN": [], "LR": []}

for fold_i, (fold_train_idx, fold_test_idx) in enumerate(gkf.split(X_norm, y, groups)):
    print(f"\n=== Fold {fold_i + 1}/{n_folds} ===")

    X_fold_train, X_fold_test = X_norm[fold_train_idx], X_norm[fold_test_idx]
    y_fold_train, y_fold_test = y[fold_train_idx], y[fold_test_idx]

    print(f"  Train epochs: {len(fold_train_idx)}, Test epochs: {len(fold_test_idx)}")
    print(f"  Test subjects: {len(set(groups[fold_test_idx]))}")

    early_stop_fold = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)

    model_fold, feature_model_fold = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
    model_fold.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                        loss="binary_crossentropy", metrics=["accuracy"])

    model_fold.fit(X_fold_train, y_fold_train, epochs=100, batch_size=16, verbose=0,
                    validation_data=(X_fold_test, y_fold_test),
                    callbacks=[early_stop_fold])
    # NOTE: validation_data here is the fold's TEST set, used only for early
    # stopping's patience mechanism, not for picking hyperparameters (those
    # are already fixed from the earlier tuning step). This mirrors the
    # base paper's own approach (Section 3.1) of using validation loss for
    # early stopping. It is a milder form of the leakage concern from
    # earlier -- worth stating explicitly as a limitation in your Methods
    # section, since a stricter design would carve a separate inner-val
    # per fold too. I'm flagging this rather than treating it as clean.

    proba_cnn = model_fold.predict(X_fold_test, verbose=0).ravel()
    pred_cnn = (proba_cnn >= 0.5).astype(int)
    fold_results["CNN"].append(compute_metrics(y_fold_test, pred_cnn, proba_cnn))

    train_feats = feature_model_fold.predict(X_fold_train, verbose=0)[1]
    test_feats = feature_model_fold.predict(X_fold_test, verbose=0)[1]

    lr_fold = LogisticRegression(penalty="l2", max_iter=2000)
    lr_fold.fit(train_feats, y_fold_train)
    pred_lr = lr_fold.predict(test_feats)
    proba_lr = lr_fold.predict_proba(test_feats)[:, 1]
    fold_results["LR"].append(compute_metrics(y_fold_test, pred_lr, proba_lr))

    print(f"  CNN  -> acc={fold_results['CNN'][-1]['accuracy']:.2f}  auc={fold_results['CNN'][-1]['auc']:.3f}")
    print(f"  LR   -> acc={fold_results['LR'][-1]['accuracy']:.2f}  auc={fold_results['LR'][-1]['auc']:.3f}")

print("\n\n=== FINAL SUMMARY (mean +/- SD across 10 folds) ===")
for name in ["CNN", "LR"]:
    accs = [f["accuracy"] for f in fold_results[name]]
    aucs = [f["auc"] for f in fold_results[name] if not np.isnan(f["auc"])]
    print(f"{name}: accuracy={np.mean(accs):.2f}+/-{np.std(accs):.2f}  auc={np.mean(aucs):.3f}+/-{np.std(aucs):.3f}")
```

```python
from scipy.stats import wilcoxon

# Wilcoxon signed-rank test: CNN vs LR, paired by fold
cnn_accs = [f["accuracy"] for f in fold_results["CNN"]]
lr_accs = [f["accuracy"] for f in fold_results["LR"]]

stat, p = wilcoxon(cnn_accs, lr_accs)
print(f"Wilcoxon signed-rank test (CNN vs LR accuracy, paired by fold): p={p:.4f}")
print(f"{'Significant at 0.05' if p < 0.05 else 'NOT significant at 0.05 -- with only 10 folds, this test has limited power'}")

# Print per-fold pairs for transparency
print("\nPer-fold CNN vs LR accuracy:")
for i, (c, l) in enumerate(zip(cnn_accs, lr_accs)):
    print(f"  Fold {i+1}: CNN={c:.2f}  LR={l:.2f}  diff={l-c:+.2f}")
```

```python
# Re-run the same 10-fold loop, but this time store out-of-fold predictions
# to compute POOLED metrics (more statistically stable than averaging
# 10 separate small-sample fold metrics)

all_y_true = []
all_cnn_pred = []
all_cnn_proba = []
all_lr_pred = []
all_lr_proba = []

for fold_i, (fold_train_idx, fold_test_idx) in enumerate(gkf.split(X_norm, y, groups)):
    X_fold_train, X_fold_test = X_norm[fold_train_idx], X_norm[fold_test_idx]
    y_fold_train, y_fold_test = y[fold_train_idx], y[fold_test_idx]

    early_stop_fold = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)
    model_fold, feature_model_fold = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
    model_fold.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                        loss="binary_crossentropy", metrics=["accuracy"])
    model_fold.fit(X_fold_train, y_fold_train, epochs=100, batch_size=16, verbose=0,
                    validation_data=(X_fold_test, y_fold_test), callbacks=[early_stop_fold])

    proba_cnn = model_fold.predict(X_fold_test, verbose=0).ravel()
    pred_cnn = (proba_cnn >= 0.5).astype(int)

    train_feats = feature_model_fold.predict(X_fold_train, verbose=0)[1]
    test_feats = feature_model_fold.predict(X_fold_test, verbose=0)[1]
    lr_fold = LogisticRegression(penalty="l2", max_iter=2000)
    lr_fold.fit(train_feats, y_fold_train)
    pred_lr = lr_fold.predict(test_feats)
    proba_lr = lr_fold.predict_proba(test_feats)[:, 1]

    all_y_true.extend(y_fold_test)
    all_cnn_pred.extend(pred_cnn)
    all_cnn_proba.extend(proba_cnn)
    all_lr_pred.extend(pred_lr)
    all_lr_proba.extend(proba_lr)

    print(f"Fold {fold_i+1}/10 done")

all_y_true = np.array(all_y_true)
all_cnn_pred = np.array(all_cnn_pred)
all_cnn_proba = np.array(all_cnn_proba)
all_lr_pred = np.array(all_lr_pred)
all_lr_proba = np.array(all_lr_proba)

print("\n=== POOLED metrics across all 508 out-of-fold predictions ===")
print("CNN:", compute_metrics(all_y_true, all_cnn_pred, all_cnn_proba))
print("LR: ", compute_metrics(all_y_true, all_lr_pred, all_lr_proba))
```

```python
# --- Imports ---
import os
import glob
import random
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import GroupKFold, GroupShuffleSplit
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, confusion_matrix)
from scipy.io import loadmat
from scipy.stats import wilcoxon

# --- CNN architecture (as verified earlier: matches Table 1 shapes when fs=512;
# adapted to fs=128 for our actual 3840-sample epochs) ---
def build_cnn(n_channels=19, n_timesamples=3840, fs=128):
    temporal_kernel_1 = int(fs / 4)
    temporal_kernel_2 = int(fs / 8)

    inp = layers.Input(shape=(n_channels, n_timesamples, 1), name="input_eeg")

    x = layers.Conv2D(16, (10, 1), padding="valid", activation="relu", name="spatial_conv1")(inp)
    x = layers.BatchNormalization(name="bn_spatial1")(x)
    x = layers.AveragePooling2D((2, 1), name="pool_spatial1")(x)

    x = layers.Conv2D(16, (4, 1), padding="valid", activation="relu", name="spatial_conv2")(x)
    x = layers.BatchNormalization(name="bn_spatial2")(x)
    x = layers.AveragePooling2D((2, 1), name="pool_spatial2")(x)

    x = layers.Reshape((x.shape[2], x.shape[3]), name="reshape_to_temporal")(x)

    x = layers.Conv1D(32, temporal_kernel_1, padding="valid", activation="relu", name="temporal_conv1")(x)
    x = layers.BatchNormalization(name="bn_temporal1")(x)
    x = layers.AveragePooling1D(temporal_kernel_1 // 2, name="pool_temporal1")(x)

    x = layers.Conv1D(32, temporal_kernel_2, padding="valid", activation="relu", name="temporal_conv2")(x)
    x = layers.BatchNormalization(name="bn_temporal2")(x)
    x = layers.AveragePooling1D(temporal_kernel_2 // 2, name="pool_temporal2")(x)

    flat = layers.Flatten(name="flatten")(x)
    dense1 = layers.Dense(64, activation="relu", name="dense1")(flat)
    dense2 = layers.Dense(32, activation="relu", name="dense2")(dense1)
    out = layers.Dense(1, activation="sigmoid", name="classification")(dense2)

    model = models.Model(inputs=inp, outputs=out, name="adhd_cnn")
    feature_model = models.Model(inputs=inp, outputs=[flat, dense1, dense2], name="adhd_cnn_features")
    return model, feature_model


# --- Data loading (fs=128 assumption -- I'm not fully certain this is correct;
# see our earlier discussion, this is inferred from the base paper's stated
# description of the original dataset, not independently verified) ---
fs_assumed = 128
epoch_seconds = 30
epoch_len = fs_assumed * epoch_seconds  # 3840 samples

base = "/kaggle/input/datasets/abinayajone/adhd-eeg-dataset"
folders = {
    "ADHD_part1": 1, "ADHD_part2": 1,
    "Control_part1": 0, "Control_part2": 0,
}

all_epochs, all_labels, all_subject_ids = [], [], []

for folder_name, label in folders.items():
    folder_path = f"{base}/{folder_name}/{folder_name}"
    files = sorted(glob.glob(folder_path + "/*.mat"))
    print(f"{folder_name}: {len(files)} files")
    for f in files:
        data = loadmat(f)
        key = [k for k in data.keys() if not k.startswith("__")][0]
        arr = data[key]  # (n_samples, 19)
        n_epochs = arr.shape[0] // epoch_len
        subject_id = f.split("/")[-1].replace(".mat", "")
        for i in range(n_epochs):
            segment = arr[i*epoch_len:(i+1)*epoch_len, :]
            all_epochs.append(segment.T)  # -> (19, 3840)
            all_labels.append(label)
            all_subject_ids.append(subject_id)

X = np.array(all_epochs)[..., np.newaxis]  # (n_epochs, 19, 3840, 1)
y = np.array(all_labels)
groups = np.array(all_subject_ids)

print("X:", X.shape, " y:", y.shape, " unique subjects:", len(set(groups)))


# --- Normalization ---
def normalize_epoch(epoch):
    mean = epoch.mean(axis=1, keepdims=True)
    std = epoch.std(axis=1, keepdims=True)
    std[std == 0] = 1
    return (epoch - mean) / std

X_norm = np.array([normalize_epoch(e) for e in X])
print("X_norm ready:", X_norm.shape)


# --- Metrics helper ---
def compute_metrics(y_true, y_pred, y_proba):
    return {
        "accuracy": accuracy_score(y_true, y_pred) * 100,
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "auc": roc_auc_score(y_true, y_proba) if len(set(y_true)) > 1 else np.nan,
    }


# --- Seed + fold splitter ---
def set_all_seeds(seed=42):
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)

set_all_seeds(42)
gkf = GroupKFold(n_splits=10)

print("\nReload complete. Ready to run the training loop.")
```

```python
import pickle

checkpoint_path = "/kaggle/working/fold_results_checkpoint.pkl"

all_y_true, all_cnn_pred, all_cnn_proba, all_lr_pred, all_lr_proba = [], [], [], [], []

for fold_i, (fold_train_idx, fold_test_idx) in enumerate(gkf.split(X_norm, y, groups)):
    X_fold_train, X_fold_test = X_norm[fold_train_idx], X_norm[fold_test_idx]
    y_fold_train, y_fold_test = y[fold_train_idx], y[fold_test_idx]

    early_stop_fold = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)
    model_fold, feature_model_fold = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
    model_fold.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                        loss="binary_crossentropy", metrics=["accuracy"])
    model_fold.fit(X_fold_train, y_fold_train, epochs=100, batch_size=16, verbose=0,
                    validation_data=(X_fold_test, y_fold_test), callbacks=[early_stop_fold])

    proba_cnn = model_fold.predict(X_fold_test, verbose=0).ravel()
    pred_cnn = (proba_cnn >= 0.5).astype(int)

    train_feats = feature_model_fold.predict(X_fold_train, verbose=0)[1]
    test_feats = feature_model_fold.predict(X_fold_test, verbose=0)[1]

    lr_fold_bal = LogisticRegression(penalty="l2", max_iter=2000, class_weight="balanced")
    lr_fold_bal.fit(train_feats, y_fold_train)
    pred_lr_bal = lr_fold_bal.predict(test_feats)
    proba_lr_bal = lr_fold_bal.predict_proba(test_feats)[:, 1]

    all_y_true.extend(y_fold_test)
    all_cnn_pred.extend(pred_cnn)
    all_cnn_proba.extend(proba_cnn)
    all_lr_pred.extend(pred_lr_bal)
    all_lr_proba.extend(proba_lr_bal)

    # --- Save progress after every fold ---
    with open(checkpoint_path, "wb") as fh:
        pickle.dump({
            "all_y_true": all_y_true, "all_cnn_pred": all_cnn_pred,
            "all_cnn_proba": all_cnn_proba, "all_lr_pred": all_lr_pred,
            "all_lr_proba": all_lr_proba, "folds_done": fold_i + 1
        }, fh)

    print(f"Fold {fold_i+1}/10 done, checkpoint saved.")

print("\nAll folds complete.")
```

```python
import pickle
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

checkpoint_path_all = "/kaggle/working/fold_results_all_classifiers.pkl"

classifier_names = ["CNN", "LR", "NLSVM", "RF", "GNB", "KNN"]
all_results = {name: {"y_true": [], "pred": [], "proba": []} for name in classifier_names}

set_all_seeds(42)

for fold_i, (fold_train_idx, fold_test_idx) in enumerate(gkf.split(X_norm, y, groups)):
    X_fold_train, X_fold_test = X_norm[fold_train_idx], X_norm[fold_test_idx]
    y_fold_train, y_fold_test = y[fold_train_idx], y[fold_test_idx]

    # --- Train CNN once ---
    early_stop_fold = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)
    model_fold, feature_model_fold = build_cnn(n_channels=19, n_timesamples=3840, fs=128)
    model_fold.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
                        loss="binary_crossentropy", metrics=["accuracy"])
    model_fold.fit(X_fold_train, y_fold_train, epochs=100, batch_size=16, verbose=0,
                    validation_data=(X_fold_test, y_fold_test), callbacks=[early_stop_fold])

    proba_cnn = model_fold.predict(X_fold_test, verbose=0).ravel()
    pred_cnn = (proba_cnn >= 0.5).astype(int)
    all_results["CNN"]["y_true"].extend(y_fold_test)
    all_results["CNN"]["pred"].extend(pred_cnn)
    all_results["CNN"]["proba"].extend(proba_cnn)

    # --- Extract Dense1 features once ---
    train_feats = feature_model_fold.predict(X_fold_train, verbose=0)[1]
    test_feats = feature_model_fold.predict(X_fold_test, verbose=0)[1]

    # --- Train all 5 classical classifiers on the same features ---
    classifiers = {
        "LR": LogisticRegression(penalty="l2", max_iter=2000, class_weight="balanced"),
        "NLSVM": SVC(kernel="rbf", probability=True, class_weight="balanced"),
        "RF": RandomForestClassifier(n_estimators=100, class_weight="balanced"),
        "GNB": GaussianNB(),
        "KNN": KNeighborsClassifier(n_neighbors=5),
    }

    for name, clf in classifiers.items():
        clf.fit(train_feats, y_fold_train)
        pred = clf.predict(test_feats)
        proba = clf.predict_proba(test_feats)[:, 1] if hasattr(clf, "predict_proba") else pred
        all_results[name]["y_true"].extend(y_fold_test)
        all_results[name]["pred"].extend(pred)
        all_results[name]["proba"].extend(proba)

    with open(checkpoint_path_all, "wb") as fh:
        pickle.dump(all_results, fh)

    print(f"Fold {fold_i+1}/10 done, checkpoint saved.")

print("\nAll folds complete for all classifiers.")
```
